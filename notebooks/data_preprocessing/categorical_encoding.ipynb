# Data preprocessing for categorical features

In this notebook, I will experiment with encoding categorical features before implementing the preprocessing logic as a reusable function in `src`.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

from telco_churn.config import CATEGORICAL_COLUMNS
from telco_churn.paths import TELCO_CUSTOMERS_TRAIN_PATH
```

## Prepare categorical features for encoding

```python
# Load training dataset
df = pd.read_csv(TELCO_CUSTOMERS_TRAIN_PATH)
df.head(3)
```

```python
# Prepare dataset with categorical columns only
df_categorical = df[list(CATEGORICAL_COLUMNS)].copy()
df_categorical.head(3)
```

```python
df_categorical.shape
```

## Encode categorical columns using OneHotEncoder

```python
encoder = OneHotEncoder(
    drop='if_binary',
    handle_unknown='ignore',
)
encoder.fit(df_categorical)
```

```python
# Transform categorical DataFrame
encoded = encoder.transform(df_categorical)
```

```python
# Create a DataFrame from encoded features
df_categorical_encoded = pd.DataFrame(
    encoded.toarray(),
    columns=encoder.get_feature_names_out(),
    index=df_categorical.index,
)
df_categorical_encoded.head(3)
```
